```python
%matplotlib inline
import matplotlib.pyplot as plt
from pathlib import Path
from netket.graph import Hypercube


FIGURES_DIR = Path("figures")
FIGURES_DIR.mkdir(parents=True, exist_ok=True)


def draw_lattice(*, length, n_dim, pbc, figsize, dpi, node_size, filename, reference_span=None):
    fig, ax = plt.subplots(figsize=figsize, dpi=dpi)
    graph = Hypercube(length=length, n_dim=n_dim, pbc=pbc)

    graph.draw(ax=ax, draw_basis_vectors=False, draw_unit_cell=False, node_size=node_size)

    if reference_span is not None and n_dim == 2:
        x0, x1 = ax.get_xlim()
        y0, y1 = ax.get_ylim()
        xc = 0.5 * (x0 + x1)
        yc = 0.5 * (y0 + y1)
        ax.set_xlim(xc - reference_span / 2, xc + reference_span / 2)
        ax.set_ylim(yc - reference_span / 2, yc + reference_span / 2)

    if n_dim == 2:
        ax.set_aspect("equal")

    ax.set_title("")
    ax.set_axis_off()
    plt.tight_layout()
    plt.show()

    fig.savefig(filename, bbox_inches="tight")
    return fig, ax, graph
```

```python
fig_4x4, ax_4x4, graph_4x4 = draw_lattice(
    length=4,
    n_dim=2,
    pbc=True,
    figsize=(4.6, 4.6),
    dpi=300,
    node_size=400,
    filename=FIGURES_DIR / "tfim_4x4_graph.png",
)

reference_span = max(
    ax_4x4.get_xlim()[1] - ax_4x4.get_xlim()[0],
    ax_4x4.get_ylim()[1] - ax_4x4.get_ylim()[0],
)
```

```python
fig_3x3, ax_3x3, graph_3x3 = draw_lattice(
    length=3,
    n_dim=2,
    pbc=True,
    figsize=(4.6, 4.6),
    dpi=300,
    node_size=400,
    filename=FIGURES_DIR / "tfim_3x3_graph.png",
    reference_span=reference_span,
)
```
